```python
import boto3
import os
from botocore import UNSIGNED
from botocore.config import Config
```

```python
bucket_name = "open-neurodata"

resource = boto3.resource("s3", config=Config(signature_version=UNSIGNED))

bucket = resource.Bucket(bucket_name)
```

```python
datasets = [
        "zebrafinch",
        'hemi',
        'fib25'
    ]

networks = [
        "baseline",
        'mtlsd',
        'lsd',
        'long_range',
        'malis',
        'aclsd',
        'acrlsd'
    ]
```

```python
out_dir_parent = "/nrs/saalfeld/heinrichl/fly_organelles/lsd/networks"
```

```python
for dataset in datasets:
    for network in networks:
        out_dir = os.path.join(out_dir_parent, dataset, network)
        os.makedirs(out_dir)
        key = f"funke/{dataset}/training/checkpoints/{network}/"

        for obj in bucket.objects.filter(Prefix=key):
            down = obj.key
            out = os.path.join(out_dir, obj.key.replace(key, ""))

            bucket.download_file(down, out)
```
